## key word arguments & default arguments
- *kwargs


```python
def summing (num1, num2, power): # parameters (3)
    sum = (num1 + num2) ** power
    return sum

#function cell
summing (1, -2, 1) # order is important

# keyword arguments = order is not important
summing (power=2, num1 =3, num2 = 1) # but there is a keyword to them as well

summing(1,-3,3)

# with default arguments
def summing (num1 = 1, num2 = 1, power = 1): # parameters (3)
    sum = (num1 + num2) ** power
    return sum
summing(3,2,2)
```

```python
#simple function for interface
def add (num1 = 1, num2 = 1):
    return num1 + num2
```

```python
import gradio as gr

#gradio interface
gr.Interface(fn=add,
    inputs=["number", "number"],
    outputs="number" # able to put text also
).launch()
```

```python
import gradio as gr

#gradio interface
gr.Interface(fn=add,
    inputs= [
        gr.Dropdown (choices = [1,2,3,4,5]),
        "number"
    ],
    outputs="number" # able to put text also
).launch()
```

```python
import gradio as gr

#gradio interface
gr.Interface(fn=add,
    inputs= [
        gr.Dropdown (choices = [1,2,3,4,5]),
        gr.Slider (minimum = 0, maximum = 100, step = 1)
    ],
    #"number" --- number field
    # "text" --- text field
    # gr.Checkbox() True False
    # gr.Image ()
    # gr.Audio ()
    # gr.file()
    outputs="number", # able to put text also
    title = "A Summing Machine!!",
    description = "This a Program for Number Only & Teaches Gradio"
).launch()
```

```python
import gradio as gr

#gradio interface
interface = gr.Interface(fn=add,
    inputs= [
        gr.Dropdown (choices = [1,2,3,4,5]),
        gr.Slider (minimum = 0, maximum = 100, step = 1)
    ],
    #"number" --- number field
    # "text" --- text field
    # gr.Checkbox() True False
    # gr.Image ()
    # gr.Audio ()
    # gr.file()
    outputs="number", # able to put text also
    title = "A Summing Machine!!",
    description = "This a Program for Number Only & Teaches Gradio"
)
interface.launch (share = True) #gradio can help us to create live link that everyone can use
```
