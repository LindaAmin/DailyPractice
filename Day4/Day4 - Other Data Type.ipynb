slicing in list

```python
# list comprehension

nums =[x for x in range(20)]
nums =[2*x for x in range(20)]
nums =[2*(x**3)+2 for x in range(20)]
print(nums)
```

```python
# slicing ---> [start:end:stepsize]

# range (start:end:stepsize)
nums[::]

nums[5:10]

nums[:5] # first 5 elements

nums[15:] # of the last 5 elements

nums[-1] # the last element

nums[-3]

nums[-3:-20:-1]

nums[::-1]

nums.reverse()
print(nums)
```

```python
name = input("Please enter your name: ")

print(f"Hello {name}! How can I help you with today's weather?")
print("I will advise you based on today's weather.")

temp = input("Please insert your temperature in Fahrenheit. Please key in 'q' or 'quit' to stop the program accordingly: ")

if temp.lower() in ["q", "quit"]:
    print("Program Stopped!")

else:
    try:
        fahrenheit = int(temp)

        celsius = (fahrenheit - 32) * 5 / 9

        if celsius < 15:
            print(f"Hello {name}, The weather is cold today. Please wear warm clothing before heading out.")

        elif celsius < 30:
            print(f"Hello {name}, The weather is pleasant today. It's a great day for outdoor activities.")

        elif celsius < 40:
            print(f"Hello {name}, The weather is hot today. It is advisable to avoid outdoor activities during peak hours.")

        else:
            print(f"Hello {name}, It's extremely hot today. Consider staying at home and avoiding unnecessary outdoor activities.")

    except ValueError:
        print("Invalid number. Please enter a correct number again.")
```
