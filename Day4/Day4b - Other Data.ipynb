## SET
- theory in math
- collection of items
- is an unordered collection of unique items
- very useful when you want to store items that should not repeat
- represent set using Venn Diagrams


```python
set_A = {1,2,3,4,4,3,2}
print(set_A)


set_B = {3,4,8,5,9,6}
print(set_B)



```

set_A | set_B # Union. Combination of both set (all emelent uniquely)

set_A & set_B # Intersection. overlap set only (comman elements in the sets)

```python
# list
names = ["idrees", "afif", "hazira", "idrees", "luqman","hazira"]

# remove all duplicates
names = list(names)
names = set(names) # Option 1
names = list(set(names)) # Option 2

print(names)
```

```python
set_A = {1,2,3,4,4,3,2}

names = {"idrees", "afif", "hazira", "idrees", "luqman","hazira"}

set_A.add(42) # to add an element
print(set_A)

set_A.update(names) # to add any iterables
print(set_A)

set_A.remove(42) # to remove element
print(set_A)

set_A.discard(42) # when not sure if element is already remove. this is not throw & error
print(set_A)

random_element = set_A.pop() # randomly returns an element
print(set_A)

set_A.clear()
print(set_A)
```
