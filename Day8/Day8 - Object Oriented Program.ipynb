### Inheritance & Polymorphism
INHERITANCE
- a new class can reuse code from a exisiting class (PARENT)
- super()._init_() # is used to call a parent class
example:
Parent ==== Animal
Children Classes

POLYMORPHISM
- the same method, gives different outputs



```python
# Parent Class
class Animal:
    def __init__(self, name, species, age):
        self.name = name
        self.species = species
        self._age = age  # protected attribute . this tells other developers, do not TOUCH FROM OUTSIDE

    def introduce(self):
        print (f"Hi. I am {self.name}\nI am a {self.species}\nI am {self.age} years old")

    def get_age(self):
        return self._age
    
    def set_age (self, new_age):
        self.age = new_age
        return f"The new age is {new_age}"
    
    def have_birthday(self):
        self._age += 1
        print(f"It's {self.name}'s birthday. Now {self._age} years old.")
```

```python
# CHILD OF ANIMAL
class Dog (Animal):

    def __init__(self, name, age): # this is what we will pass ourselves
        # super () refers to the parentsclass --- could have used Animal as well
        self._age = age
        super().__init__ (name, "Dog", age)
    
    def sound(self):
        print (f" {self.name} says: Woof!!!")
    
    def detect_drugs (self):
        if self._age >5:
            return f"I will sniff out any drugs"
        else:
            return f"{self.name} say. I am just {self.age} old. You are ok?"
```

```python
# examples of working with a class
dog1 = Dog("Johnny", 5)
```
