# Recursion
👉 Function calling itself again and again
👉 Stops when a condition is met

Every recursive function must have two parts:
1. A base case - A condition that stops the recursion
2. A recursive case - The function calling itself with a modified argument. do a small part & call function

```python
# Basis coding for function

def function (parameters):
    if <condition>:
        return result
    else:
        return function (<modified parameters)
```

```python
# 10 to 1
for i in range(10, 0, -1):
    print(i)


def countdown(n):
  # base case
  if n == 0:
    print("Done!")
  else:
    print(n)
    countdown(n - 1)

countdown(5)
```

```python
# 5 → 4 → 3 → 2 → 1 (going down)
# then multiply going back up

def factorial(n):
  # Base case
  if n == 0 or n == 1:
    return 1
  # Recursive case
  else:
    return n * factorial(n - 1)

print(factorial(5))
```

```python
# factorial using for loop

def multiplyn (n): # parameter
    factorial = 1
    for i in range (1, n+1):
        factorial = factorial * i
    return factorial

multiplyn(3) # argument
```

```python
# factorial using for loop

def multiplyn (n): # parameter
    # base case
    if n == 1:
        return n

    # recursive
    return n * multiplyn (n-1)

multiplyn(10) # argument
```

```python
# create a timer function by using recursion
# start from 10 go to 0

def timer(n):
    print(f"Time:  {n}")

    if n == 0:
        print("Time's up!")
    else:
        timer(n - 1)
    
timer(10) # start timer
```

```python
# create a timer function by using recursion
# start from 10 go to 0

import time

timer = 5

for i in range(timer):
    time.sleep(1)
    print(f"Time:{i+1}")

#rocket timers do a countdown TimeoutError
for i in range(timer,0,-1):
    time.sleep(1)
    print(f"Time:{i+1}")
```

```python
# Recursive case
def countdown(n):
    print(n)

    if n == 0:
        print("Done!")
    else:
        countdown(n - 1)

# function call
countdown(5)
```

```python
def fibonacci(n):
  if n <= 1:
    return n
  else:
    return fibonacci(n - 1) + fibonacci(n - 2)

print(fibonacci(7))
```

```python
# Recursion with Lists
# Recursion can be used to process lists by handling one element at a time

def sum_list(numbers):
  if len(numbers) == 0:
    return 0
  else:
    return numbers[0] + sum_list(numbers[1:])

my_list = [1, 2, 3, 4, 5]
print(sum_list(my_list))


def find_max(numbers):
  if len(numbers) == 1:
    return numbers[0]
  else:
    max_of_rest = find_max(numbers[1:])
    return numbers[0] if numbers[0] > max_of_rest else max_of_rest

my_list = [3, 7, 2, 9, 1]
print(find_max(my_list))
```
