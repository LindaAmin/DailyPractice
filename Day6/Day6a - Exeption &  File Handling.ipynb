## file handling:
    - how does python access a file
    - operations like:
        - read
        - write
        - open
        - close

"r" - Read - Default value. Opens a file for reading, error if the file does not exist

"a" - Append - Opens a file for appending, creates the file if it does not exist

"w" - Write - Opens a file for writing, creates the file if it does not exist

"x" - Create - Creates the specified file, returns an error if the file exists

```python
# to open a file
file = open("linda.txt", "r") # read
```

```python
try:
    file = open("linda.txt", "r")
except Exception as e:
    print("Something is error",e)
```

```python
# create a file
file = open("linda.csv", "w") # write
```

```python
# (1) write on a file
# open in a write mood
file = open("linda.txt", "w+") # open in write & read mood

file.write("name, age, phone, subject\n")
```

```python
# (2)
file.read()
```

```python
# (3) append on file
file = open("linda.csv", "a") # open in append mood
```

contact manager - using the "with"
- with open (<filename>, "r") as file:
    file.read()
    file.write()
    file.close # is done automatically & you dont need close the files once use content
- automatically closes

```python
#\n is for break the sentences

with open("linda.txt", "r+") as f:
    f.write("linda is awesome\n")
    f.write("linda is good\n")
    content = f.read()
    print(content)
```

```python
#\n is for break the sentences

with open("linda.txt", "r+") as f:
    f.write("linda is awesome\n")
    f.write("linda is good\n")
```
